```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import seaborn.objects as so
```

```python
df = pd.concat([pd.read_csv('data/Fraudulent_E-Commerce_Transaction_Data.csv'), pd.read_csv('data/Fraudulent_E-Commerce_Transaction_Data_2.csv')], ignore_index=True)
df
```

```python
df.info()
```

```python
df.drop_duplicates(inplace=True)
df.drop(columns=['Transaction ID', 'Customer ID', 'IP Address'], inplace=True)
```

```python
df.describe()
```

```python
for col in df.columns:
    print(f"\n{df[col].value_counts()}\n")
```

```python
# chuyển transaction date thành các cột year, month, day, hour, minute, second
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])
df['TransMonth'] = df['Transaction Date'].dt.month
df['TransDay'] = df['Transaction Date'].dt.day
df['TransTime'] = df['Transaction Date'].dt.time
df
```

```python
# chỉnh customer location bỏ East, West, North, South, Lake, Central, gom về chung location
df['Customer Location'] = df['Customer Location'].str.replace('East|West|North|South', '', regex=True).str.strip()
df['Customer Location']
```

```python
df
```

```python
#pivot table location và product category
pivot_table = pd.pivot_table(df, values='Transaction Amount', index='Customer Location', columns='Product Category', aggfunc='sum', fill_value=0).astype(float).round(2)
pivot_table
```
